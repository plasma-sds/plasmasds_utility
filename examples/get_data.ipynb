# Getting data with plasmasds_utility

This notebook gets a data file from the plasma-sds data server.
It uses renate's public test file, so it works without an SSH key.
The same example for a terminal is `get_data.py`.

A client package creates one client, named after its folder on the server.
`client_dir()` shows where its data is kept on this computer.

````python
import plasmasds_utility
from plasmasds_utility import DataClient

data = DataClient("renate-od")
data.client_dir()
````

`get()` returns the local path of a file, downloading it first if it is missing.
The first call in a session also prints the options for checking the server for newer data.

`private=False` takes public data only.
Client packages normally call `data.get(key)`, which takes private data if you have access (an SSH key, set once with `plasmasds_utility.set_ssh_key("~/.ssh/your_key")`) and public data otherwise.

````python
key = "test_dataset/access_tests/public_test.txt"
path = data.get(key, private=False)
path, path.read_text(encoding="utf-8")
````

Calling it again only checks that the file is on disk.
To download it again if the server has a newer version, use `update="if_newer"`; `data.check_updates()` does this for every local file of the client.

````python
data.get(key, private=False, update="if_newer")
````

To keep a client's data in another folder, for example on a larger disk, save it once:

```python
data.set_working_dir("/path/to/folder")
```

It is not run here, as it changes your saved settings.

With the default `get(key)`, files you cannot get as private data come from public data, with a warning.
This lists them; here there are none, as public data was asked for.

````python
plasmasds_utility.show_public_fallbacks()
````

If something does not work, `plasmasds_utility.check_connection()` checks both ways to the server and shows the settings in use.
